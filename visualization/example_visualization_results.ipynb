## Confusion matrices per agent

Reads the `*_result_matrix.out` file: one NxN confusion matrix block per agent (blocks separated by blank lines), in the same agent order as `*_heads.out`.

```python
import glob

def read_confusion_matrices(path):
    """Parse a *_result_matrix.out file into a list of NxN confusion matrices, one per agent."""
    with open(path, mode='r', encoding='utf-8-sig') as f:
        content = f.read()
    blocks = [b for b in content.strip().split('\n\n') if b.strip()]
    return [np.array([[int(x) for x in row.split()] for row in b.strip().split('\n')]) for b in blocks]

result_matrix_file = sorted(glob.glob(file_path + '/*_result_matrix.out'))[0]
confusion_matrices = read_confusion_matrices(result_matrix_file)
num_classes = confusion_matrices[0].shape[0]
print(f'Loaded {len(confusion_matrices)} confusion matrices of shape {confusion_matrices[0].shape} from {result_matrix_file}')
```

```python
#VISUALIZE CONFUSION MATRIX GRID, ONE SUBPLOT PER AGENT (row-normalized = recall)
n_cols = 4
n_rows = int(np.ceil(num_agents / n_cols))
fig, axs = plt.subplots(n_rows, n_cols, figsize=(3.2 * n_cols, 3.2 * n_rows))
axs = np.atleast_1d(axs).flatten()

for agent_id in range(num_agents):
    cm_counts = confusion_matrices[agent_id]
    cm_norm = cm_counts / cm_counts.sum(axis=1, keepdims=True)
    ax = axs[agent_id]
    im = ax.imshow(cm_norm, cmap='viridis', vmin=0, vmax=1)
    ax.set_title(f'Agent {agent_id}', fontsize=9)
    ax.set_xlabel('Predicted', fontsize=7)
    ax.set_ylabel('True', fontsize=7)
    ax.set_xticks(range(num_classes))
    ax.set_yticks(range(num_classes))
    ax.tick_params(labelsize=6)
    for i in range(num_classes):
        for j in range(num_classes):
            ax.text(j, i, cm_counts[i, j], ha='center', va='center', fontsize=5,
                    color='white' if cm_norm[i, j] > 0.5 else 'black')

for ax in axs[num_agents:]:
    ax.axis('off')

fig.colorbar(im, ax=axs.tolist(), shrink=0.6, label='Recall (row-normalized)')
fig.suptitle(f'Confusion matrices per agent — {run_name}')
plt.savefig(save_path + 'agents_confusion_matrices.pdf', bbox_inches='tight')
plt.show()
```

```python
def plot_confusion_matrix(cm_counts, agent_id, normalize=True, ax=None, cmap='viridis'):
    """Plot a single agent's confusion matrix full-size, for closer inspection."""
    if ax is None:
        fig, ax = plt.subplots(figsize=(5, 5))
    data = cm_counts / cm_counts.sum(axis=1, keepdims=True) if normalize else cm_counts
    im = ax.imshow(data, cmap=cmap, vmin=0, vmax=1 if normalize else data.max())
    ax.set_title(f'Agent {agent_id}')
    ax.set_xlabel('Predicted label')
    ax.set_ylabel('True label')
    n = cm_counts.shape[0]
    ax.set_xticks(range(n))
    ax.set_yticks(range(n))
    for i in range(n):
        for j in range(n):
            txt = f'{data[i, j]:.2f}' if normalize else f'{cm_counts[i, j]}'
            ax.text(j, i, txt, ha='center', va='center', fontsize=8,
                    color='white' if data[i, j] > data.max() / 2 else 'black')
    plt.colorbar(im, ax=ax, shrink=0.8)
    return ax

# Example: inspect a single agent close-up
plot_confusion_matrix(confusion_matrices[0], agent_id=0)
plt.show()
```

## Data partitions per agent

Reads the `*_partitions.out` file: one AgentsxClasses instance-count block per split (train, then the shared test set), separated by a blank line.

```python
def read_partitions(path):
    """Parse a *_partitions.out file into a list of AgentsxClasses instance-count blocks (train, test, ...)."""
    with open(path, mode='r', encoding='utf-8-sig') as f:
        content = f.read()
    blocks = [b for b in content.strip().split('\n\n') if b.strip()]
    return [np.array([[int(x) for x in row.split()] for row in b.strip().split('\n')]) for b in blocks]

partitions_file = sorted(glob.glob(file_path + '/*_partitions.out'))[0]
partition_blocks = read_partitions(partitions_file)
train_partition = partition_blocks[0]
test_partition = partition_blocks[1] if len(partition_blocks) > 1 else None
num_partition_classes = train_partition.shape[1]
print(f'Loaded partitions from {partitions_file}: train {train_partition.shape}' +
      (f', test {test_partition.shape}' if test_partition is not None else ''))
```

```python
#VISUALIZE PARTITIONS HEATMAP: ROWS = AGENTS, COLUMNS = CLASS INSTANCE COUNT
blocks_to_plot = [('Train partition', train_partition)]
if test_partition is not None:
    blocks_to_plot.append(('Test partition', test_partition))

fig, axs = plt.subplots(1, len(blocks_to_plot), figsize=(6 * len(blocks_to_plot), 4.5))
axs = np.atleast_1d(axs)

for ax, (title, block) in zip(axs, blocks_to_plot):
    im = ax.imshow(block, cmap='viridis', aspect='auto')
    ax.set_title(f'{title} — {run_name}', fontsize=10)
    ax.set_xlabel('Class')
    ax.set_ylabel('Agent ID')
    ax.set_xticks(range(num_partition_classes))
    ax.set_yticks(range(num_agents))
    vmax = block.max()
    for i in range(num_agents):
        for j in range(num_partition_classes):
            ax.text(j, i, block[i, j], ha='center', va='center', fontsize=7,
                    color='white' if block[i, j] < vmax / 2 else 'black')
    plt.colorbar(im, ax=ax, shrink=0.8, label='Instance count')

plt.tight_layout()
plt.savefig(save_path + 'agents_data_partitions.pdf', bbox_inches='tight')
plt.show()
```

## Accuracy & Macro F1 evolution

Final-round accuracy and macro F1 per agent are shown together in `accuracy_f1_by_agent.pdf`. This section covers the full evolution across communication rounds, per agent, from the `**acc_avg` and `**macro_f1` lines in `*_raw.out`.

```python
#PARSE MACRO F1 EVOLUTION FROM raw.out ('**macro_f1: (0, v0) (1, v1) ...'), split per agent round-robin (mirrors the acc_avg parsing above)
macro_f1_line = next(l for l in raw_tplgys[0] if l.startswith('**macro_f1'))
tmp_f1 = macro_f1_line.split()
f1_values = [float(tok[:-1]) for tok in tmp_f1[2::2]]  # value tokens, e.g. '0.1735)'
f1_values = f1_values[1:]  # drop round 0 (pre-training eval), align with cid round numbering

f1_agnt = [[] for _ in range(num_agents)]
for round_idx, val in enumerate(f1_values, start=1):
    f1_agnt[(round_idx - 1) % num_agents].append(val)

min_len = min(len(series) for series in f1_agnt)
avg_f1 = np.mean([series[:min_len] for series in f1_agnt], axis=0)
print(f'{len(f1_values)} recorded rounds, {min_len} full cycles per agent')
```

```python
#VISUALIZE MACRO F1 EVOLUTION PER AGENT ACROSS COMM ROUNDS
font = {'size': 9}
plt.rc('font', **font)
viridis = cm.get_cmap('viridis', num_agents)
for agent_id in range(num_agents):
    plt.plot(f1_agnt[agent_id][:min_len], color=viridis(agent_id / max(num_agents - 1, 1)), alpha=0.5, label=f'Agent {agent_id}')
plt.plot(avg_f1, color='black', linewidth=2, label='Mean')
plt.ylim(0, 1.)
plt.xlabel('Comm Round')
plt.ylabel('Macro F1')
plt.title(f'Macro F1 evolution — {run_name}')
plt.legend(fontsize=6, ncol=2, loc='lower right')
plt.grid()
plt.savefig(save_path + 'agents_f1_by_round.pdf', bbox_inches='tight')
plt.show()
```

```python
#PARSE ACCURACY EVOLUTION FROM raw.out ('**acc_avg: (0, v0) (1, v1) ...'), split per agent round-robin (mirrors the macro_f1 parsing above)
acc_evo_line = next(l for l in raw_tplgys[0] if l.startswith('**acc_avg'))
tmp_acc_evo = acc_evo_line.split()
acc_evo_values = [float(tok[:-1]) for tok in tmp_acc_evo[2::2]]  # value tokens, e.g. '0.1735)'
acc_evo_values = acc_evo_values[1:]  # drop round 0 (pre-training eval), align with cid round numbering

acc_evo_agnt = [[] for _ in range(num_agents)]
for round_idx, val in enumerate(acc_evo_values, start=1):
    acc_evo_agnt[(round_idx - 1) % num_agents].append(val)

min_len_evo = min(len(series) for series in acc_evo_agnt)
avg_acc_evo = np.mean([series[:min_len_evo] for series in acc_evo_agnt], axis=0)
print(f'{len(acc_evo_values)} recorded rounds, {min_len_evo} full cycles per agent')
```

```python
#VISUALIZE ACCURACY EVOLUTION PER AGENT ACROSS COMM ROUNDS
font = {'size': 9}
plt.rc('font', **font)
viridis = cm.get_cmap('viridis', num_agents)
for agent_id in range(num_agents):
    plt.plot(acc_evo_agnt[agent_id][:min_len_evo], color=viridis(agent_id / max(num_agents - 1, 1)), alpha=0.5, label=f'Agent {agent_id}')
plt.plot(avg_acc_evo, color='black', linewidth=2, label='Mean')
plt.ylim(0, 1.)
plt.xlabel('Comm Round')
plt.ylabel('Acc')
plt.title(f'Accuracy evolution — {run_name}')
plt.legend(fontsize=6, ncol=2, loc='lower right')
plt.grid()
plt.savefig(save_path + 'agents_accuracy_by_round.pdf', bbox_inches='tight')
plt.show()
```
